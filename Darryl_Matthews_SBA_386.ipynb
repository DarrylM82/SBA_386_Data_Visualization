### 6. Print the "total_profit" and "month_number" columns only

```python
# subset of only the two requested columns
print(df[['total_profit', 'month_number']])
```

### 7. Total profit of all months shown using a bar plot
- X label = Month number
- Y label = Total profit
- Title = Company profit per month

```python
# bar plot of total profit per month
plt.bar(df['month_number'], df['total_profit'])
plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.title('Company profit per month')
plt.grid(linestyle='--')
plt.show()
```

### 8. Total profit of all months shown using a line plot
- Dotted line style, red line color
- Legend at the lower right
- X label = Month number, Y label = Total profit
- Title = Company profit per month
- Circle marker, marker color = black
- Line width = 3

```python
# styled line plot of total profit per month
# instructions specify a dotted line (':'); the reference image on the assignment appears dashed, I just stuck with what was orginally called out
plt.plot(
    df['month_number'], df['total_profit'],
    label='Profit data of last year',
    color='r',
    linestyle=':',
    linewidth=3,
    marker='o',
    markerfacecolor='k',
    markeredgecolor='k'
)
plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.title('Company profit per month')
plt.xticks(df['month_number'])
plt.yticks([100000, 200000, 300000, 400000, 500000])
plt.legend(loc='lower right')
plt.show()
```

### 9. Print all of the product sales data and show it using a multi-line plot
One line per product, each with its own color, plus a legend.

```python
# print every product's monthly sales
products = ['facecream', 'facewash', 'toothpaste', 'bathingsoap', 'shampoo', 'moisturizer']
print(df[['month_number'] + products])
```

```python
# multi-line plot: one line (with circle markers) per product
for product in products:
    plt.plot(df['month_number'], df[product], marker='o', linewidth=2, label=product + ' Sales Data')

plt.xlabel('Month Number')
plt.ylabel('Sales units in number')
plt.title('Sales data')
plt.xticks(df['month_number'])
plt.yticks([1000, 2000, 4000, 6000, 8000, 10000, 12000, 15000, 18000])
plt.legend(loc='upper left')
plt.show()
```

### 10. Bathing soap sales data for each month shown using a scatter plot
- Grid added with a dashed line style

- Grid added with a dashed line style ('--')
- Note: the written instruction shows "-", but the reference image shows a dashed grid and the instructor confirmed in class it should be "--", so I used a dashed grid instead per instruction.

```python
# scatter plot of bathing soap sales per month
plt.scatter(df['month_number'], df['bathingsoap'], label='bathing soap Sales data')
plt.xlabel('Month Number')
plt.ylabel('Number of units Sold')
plt.title('bathingsoap Sales data')
plt.xticks(df['month_number'])
# grid uses '--' (dashed) to match the reference screenshot and instructor clarification;
# the written instruction lists "-", which would draw a solid grid
plt.grid(True, linewidth=1, linestyle='--')
plt.legend(loc='upper left')
plt.show()
```

   **Observation:** Bathing soap sales were lowest in month 2 (6,100 units) and highest in month 12 (14,400 units). Sales climbed steadily over the last quarter, from 8,100 in month 9 to 14,400 in month 12.

# Section Two

### 1. Line chart of date versus temperature
- X label = Date
- Y label = temperature
- Title = Date-wise Temperature

```python
# given dataset
date = ['25/12', '26/12', '27/12']
temp = [8.5, 10.5, 6.8]

# line chart of date vs temperature
plt.plot(date, temp)
plt.xlabel('Date')
plt.ylabel('temperature')
plt.title('Date-wise Temperature')
plt.show()
```

**Observation:** Temperature rose from 8.5 on 25/12 to a high of 10.5 on 26/12, then dropped to 6.8 on 27/12, the lowest of the three days. The dates are plotted as text labels, so they appear in the order given.

### 2. Average weight against average height
- Dash-dot line style, green line color
- Legend at the lower right
- X label = Weight in kg, Y label = Height in cm
- Title = Average weight with respect to average height
- Circle marker, marker size = 10, marker color = green

```python
# given dataset
height = [121.9, 124.5, 129.5, 134.6, 139.7, 147.3, 152.4, 157.5, 162.6]
weight = [19.7, 21.3, 23.5, 25.9, 28.5, 32.1, 35.7, 39.6, 43.2]

# styled line plot of weight vs height
# weight on the x-axis and height on the y-axis, matching the required labels and the reference image
plt.plot(
    weight, height,
    color='green',
    linestyle='-.',
    marker='o',
    markersize=10,
    markerfacecolor='green',
    label='Average weight vs height'
)
plt.xlabel('Weight in kg')
plt.ylabel('Height in cm')
plt.title('Average weight with respect to average height')
plt.legend(loc='lower right')
plt.show()
```

**Observation:** Height increases steadily as weight increases, showing a strong positive relationship. Average height rises from 121.9 cm at 19.7 kg to 162.6 cm at 43.2 kg.
